```python
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, StratifiedKFold
```

```python
dataset_dir = "/content/drive/MyDrive/FDL Project/data/flower_photos"

# Check if the dataset directory exists
if not os.path.exists(dataset_dir):
    raise FileNotFoundError(
        f"The directory '{dataset_dir}' does not exist. "
        "Please ensure the dataset has been extracted correctly. "
        "You might need to run the `tar` command in the previous cell to extract 'flower_photos.tgz'."
    )

image_paths = []
labels = []

for class_label in os.listdir(dataset_dir):
    class_path = os.path.join(dataset_dir, class_label)
    if os.path.isdir(class_path):
        for img_name in os.listdir(class_path):
            image_paths.append(os.path.join(class_path, img_name))
            labels.append(class_label)

image_paths = np.array(image_paths)
labels = np.array(labels)

cv_images, test_images, cv_labels, test_labels = train_test_split(
    image_paths, labels, test_size=0.20, stratify=labels, random_state=42
)

print(f"Total images found: {len(image_paths)}")
print(f"Images for Cross-Val: {len(cv_images)} | Permanent Test Images: {len(test_images)}")
```
